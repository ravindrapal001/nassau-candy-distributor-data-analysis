```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load Data
df=pd.read_csv("nassau_candy.csv")
df.head()
```

```python
# Shape
print("Row:",df.shape[0])
print("Columns:",df.shape[1])
```

```python
# Information
df.info()
```

```python
# Missing Values
missing=df.isnull().sum()
missing[missing>0]
```

```python
# Duplicate rows
print("Duplicate rows:",df.duplicated().sum())
```

```python
# Unique Values
print("Unique Order IDs:",df["Order ID"].nunique())
print("Unique Products:",df["Product Name"].nunique())
print("Unique States:",df["State/Province"].nunique())
print("Unique Citites:",df["City"].nunique())
print("Unique Customers:",df["Customer ID"].nunique())
```

```python
# Date Inspection
df["Order Date"]=pd.to_datetime(df["Order Date"], dayfirst=True, errors="coerce")
df["Ship Date"]=pd.to_datetime(df["Ship Date"], dayfirst=True, errors="coerce")

print("Order Date Range:")
print(df["Order Date"].min(),"to",df["Order Date"].max())

print("\nShip Date Range:")
print(df["Ship Date"].min(),"to",df["Ship Date"].max())
```

```python
# Numerical Summary
df[["Sales","Units","Gross Profit","Cost"]].describe()
```

```python
# Business Totals
print("Total Sales:",df["Sales"].sum())
print("Total Units:",df["Units"].sum())
print("Total Gross Profit:",df["Gross Profit"].sum())
print("Total Cost:",df["Cost"].sum())
```

```python
# Profit Validation
profit_difference=(df["Sales"]-df["Cost"]-df["Gross Profit"])
print("Profit mismatches:",(profit_difference.abs()>1e-9).sum())
```

```python
# Initial Python Business Profiling
ship_mode_summary=(df.groupby("Ship Mode").agg(Shipments=("Order ID","count"),Sales=("Sales","sum"),Units=("Units","sum"),Gross_Profits=("Gross Profit","sum")).sort_values("Shipments",ascending=False))
ship_mode_summary
```

```python
region_summary=(df.groupby("Region").agg(Shipments=("Order ID","count"),Sales=("Sales","sum"),Units=("Units","sum"),Gross_Profits=("Gross Profit","sum")).sort_values("Shipments",ascending=False))
region_summary
```

```python
# Lead-Time KPIs
lead_time=(df["Ship Date"]-df["Order Date"]).dt.days
lead_time.describe()
```
