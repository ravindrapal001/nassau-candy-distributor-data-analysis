```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("nassau_candy.csv")
df.head()
```

```python



# Required columns available in the original dataset
required = [
    "Row ID",
    "Order ID",
    "Order Date",
    "Ship Date",
    "State/Province",
    "Ship Mode",
    "Sales",
    "Gross Profit"
]

# Check for missing required columns
missing_columns = [
    c for c in required if c not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

# Convert date columns into datetime format
df["Order_Date"] = pd.to_datetime(
    df["Order Date"], errors="coerce"
)

df["Ship_Date"] = pd.to_datetime(
    df["Ship Date"], errors="coerce"
)

# Calculate shipping lead time in days
df["Shipping_Lead_Time_Days"] = (
    df["Ship_Date"] - df["Order_Date"]
).dt.days

# Validate the dataset
print("Rows:", len(df))

print(
    "Invalid dates:",
    df["Order_Date"].isna().sum()
    + df["Ship_Date"].isna().sum()
)

print(
    "Negative lead times:",
    (df["Shipping_Lead_Time_Days"] < 0).sum()
)

# Display actual available columns
print("\nAvailable columns:")
print(df.columns.tolist())

# Preview the validated data
display(df.head())
```

```python



# Required columns available in the original dataset
required = [
    "Row ID",
    "Order ID",
    "Order Date",
    "Ship Date",
    "State/Province",
    "Ship Mode",
    "Sales",
    "Gross Profit"
]

# Check for missing required columns
missing_columns = [
    c for c in required if c not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

# Convert date columns into datetime format
df["Order_Date"] = pd.to_datetime(
    df["Order Date"], errors="coerce"
)

df["Ship_Date"] = pd.to_datetime(
    df["Ship Date"], errors="coerce"
)

# Calculate shipping lead time in days
df["Shipping_Lead_Time_Days"] = (
    df["Ship_Date"] - df["Order_Date"]
).dt.days

# Validate the dataset
print("Rows:", len(df))

print(
    "Invalid dates:",
    df["Order_Date"].isna().sum()
    + df["Ship_Date"].isna().sum()
)

print(
    "Negative lead times:",
    (df["Shipping_Lead_Time_Days"] < 0).sum()
)

# Display actual available columns
print("\nAvailable columns:")
print(df.columns.tolist())

# Preview the validated data
display(df.head())
```

```python

# Step 1: Filter valid records
valid = df.loc[
    df["Order_Date"].notna()
    & df["Ship_Date"].notna()
    & df["Shipping_Lead_Time_Days"].notna()
    & (df["Shipping_Lead_Time_Days"] >= 0)
    & df["City"].notna()
    & df["State/Province"].notna()
].copy()

# Step 2: Create a geographic group
valid["Route"] = (
    valid["City"].astype(str).str.strip()
    + " -> "
    + valid["State/Province"].astype(str).str.strip()
)

# Step 3: Calculate geographic-group metrics
route_metrics = (
    valid.groupby("Route")
    .agg(
        Record_Count=("Row ID", "count"),
        Distinct_Orders=("Order ID", "nunique"),
        Mean_Lead_Time=("Shipping_Lead_Time_Days", "mean"),
        Median_Lead_Time=("Shipping_Lead_Time_Days", "median"),
        Lead_Time_Std=("Shipping_Lead_Time_Days", "std"),
        P90_Lead_Time=(
            "Shipping_Lead_Time_Days",
            lambda x: x.quantile(0.90)
        ),
        Sales=("Sales", "sum"),
        Gross_Profit=("Gross Profit", "sum")
    )
    .reset_index()
)

# Step 4: Round lead-time metrics
metric_columns = [
    "Mean_Lead_Time",
    "Median_Lead_Time",
    "Lead_Time_Std",
    "P90_Lead_Time"
]

route_metrics[metric_columns] = (
    route_metrics[metric_columns].round(2)
)

# Step 5: Sort by record volume
route_metrics = route_metrics.sort_values(
    "Record_Count",
    ascending=False
)

# Step 6: Display the results
display(route_metrics.head(10))
```

```python

# Define a minimum record threshold only if justified
# by the project brief or supervisor.
MIN_RECORDS = None

# Use the geographic-group metrics calculated earlier
eligible_routes = route_metrics.copy()

# Apply the threshold only when one is specified
if MIN_RECORDS is not None:
    eligible_routes = eligible_routes.loc[
        eligible_routes["Record_Count"] >= MIN_RECORDS
    ].copy()

# Identify geographic groups with the shortest median lead time
fastest_routes = (
    eligible_routes
    .sort_values(
        ["Median_Lead_Time", "Record_Count"],
        ascending=[True, False]
    )
    .head(10)
)

# Identify geographic groups with the longest median lead time
slowest_routes = (
    eligible_routes
    .sort_values(
        ["Median_Lead_Time", "Record_Count"],
        ascending=[False, False]
    )
    .head(10)
)

print("10 Geographic Groups with the Shortest Median Lead Time")
display(fastest_routes)

print("10 Geographic Groups with the Longest Median Lead Time")
display(slowest_routes)
```

```python

from pathlib import Path

# Create the output directory for Day 06
output_dir = Path("outputs/day_06")
output_dir.mkdir(parents=True, exist_ok=True)

# Save geographic-group efficiency metrics
route_metrics.to_csv(
    output_dir / "geographic_group_efficiency_metrics.csv",
    index=False
)

# Save the 10 geographic groups with the shortest median lead time
fastest_routes.to_csv(
    output_dir / "top_10_fastest_geographic_groups.csv",
    index=False
)

# Save the 10 geographic groups with the longest median lead time
slowest_routes.to_csv(
    output_dir / "top_10_slowest_geographic_groups.csv",
    index=False
)

print("Day 06 benchmark outputs saved to:", output_dir.resolve())
print("Files created:")
print("- geographic_group_efficiency_metrics.csv")
print("- top_10_fastest_geographic_groups.csv")
print("- top_10_slowest_geographic_groups.csv")
```
